```python
import pandas as pd
import numpy as np

data = {
    "source":  ["web", "web", "store", "app", "app", "web", "store", "web", "app", "app"],
    "amount":  [250.0, 1300.0, 90.5, np.nan, 640.0, 180.0, 415.0, np.nan, 2100.0, 320.0],
    "items":   [1, 3, 2, 3, 2, 3, 1, 1, 2, 1],
    "express": ["yes", "yes", "no", "yes", "no", "no", "no", "yes", "no", "no"],
}

labels = ["ord_1", "ord_2", "ord_3", "ord_4", "ord_5",
          "ord_6", "ord_7", "ord_8", "ord_9", "ord_10"]
```

```python
#Завдання 1. Створення таблиці
orders = pd.DataFrame(data) #перетворює словник на датафрейм
print(orders) # друкує таблицю

orders = pd.DataFrame(data, index = labels) #перетворення словника на таблицю з явним указанням, що взяти за індекси
print(orders) #друкує оновлену таблицю
print('Розмір таблиці:', orders.shape) #виводить кількість рядків та кількість стовпців
print(orders.dtypes) #виводить типи даних кожної колонки
```

```python
#Завдання 2. Перший погляд на дані
print(orders.head(3)) # виводить перші три рядки таблиці
print(orders.tail(2)) # виводить останні два рядки таблиці
print('В колонці amount непорожніх значень',orders['amount'].count()) # повертає кількість непорожніх значень колонці amount
print('Порожніх значень в кожній колонці:''\n',orders.isna().sum()) # метод isna перевіряє чи є значення пустим, повертаючи 1 чи 0, а sum підбиває результат
print(orders.describe().loc[['count', 'mean', 'min', 'max', '25%', '50%', '75%']]) # виводить тільки вказані значення з методу describe
```

```python
#3. Вибірка колонок
print(orders[['source']]) # виводить колонку source
print(orders[['source', 'amount']]) # виводить разом колонки source та amount
print('Назви колонок таблиці:''\n', orders.columns) # виводить назви всіх колонок в таблиці списком
print('Унікальні значення колонки source:',orders['source'].unique()) # виводить унікальі значення з колонки source
```

```python
#Завдання 4. Рядки за позицією
print(orders.iloc[[0]]) # виводить найперший рядок таблиці
print(orders.iloc[[0,2,3]]) # виводить рядки з позиціями 0, 2 та 3
print(orders.iloc[[0,2,3], [0,1]]) # виводить рядки з позиціями 0, 2 та 3 та стовпці 0 та 1
print(orders[1:4]) # виводить рядки з позиціями 1 - 3
```

```python
#Завдання 5. Рядки за міткою
print('Рядок ord_5 колонки source:',orders.loc['ord_5', 'source']) # виводить значення, яке знаходиться в таблиці у рядку з міткою ord_5 у стовпці source
print(orders.loc[['ord_1', 'ord_3', 'ord_6']]) # виводить рядки з мітками ord_1, ord_3 та ord_6.
print(orders.loc[['ord_1', 'ord_3', 'ord_6'], ['source', 'items']]) # виводить рядки з мітками ord_1, ord_3 та ord_6 по стовпцях source та items
```

```python
# Завдання 6. Фільтрація за однією умовою
print(orders[orders['amount'] > 500]) # виводить рядки, в яких значення по стовпцю amount > 500
print(orders[orders['source'] == 'web']) # виводить рядки, в яких джерело замовення - web
print(orders[orders['express'] == 'yes']) # виводить рядки зі швидкою доставкою
print(orders[orders['source'] != 'app'])  # виводить рядки, в яких джерело замовення - не app
print('Кількість рядків із джерелом замовлення відмінних від app:',len(orders[orders['source'] != 'app'])) # виводить кількість рядків в яких джерело замолвлення не app
```

```python
#Завдання 7. Фільтрація за кількома умовами
print(orders[(orders['source'] == 'app') & (orders['items'] > 1)]) #виводить замовлення з каналу app, у яких items більше за 1
print(orders[(orders['express'] == 'no') & (orders['amount'] >= 400)]) #виводить замовлення без швидкої доставки, у яких amount більша або дорівнює 400
print(orders[(orders['amount'] < 300) | (pd.isna(orders['amount']))]) #виводить замовлення, де amount менша за 300 або відсутня
print(orders[orders['source'].isin(['web', 'store'])]) #виводить замовлення, канал яких в переліку web або store
print(orders[orders['items'].between(2, 3)]) #виводить замовлення, де items у діапазоні від 2 до 3 включно
print(orders[orders['amount'].between(200, 700)]) #виводить замовлення, де amount у діапазоні від 200 до 700 включно
print(orders.loc[orders['items'] == 1, ['source', 'items']]) #виводить тільки колонки source та items для замовлень, у яких items дорівнює 1
```

```python
# Завдання 8. Пропущені значення
print(orders.loc[orders["amount"].notna()]) #через метод loc визначаємо, які рядки потрібно вивести: а саме, ті, в яких amount  заповнена
print(orders.loc[orders["amount"].isna()]) # виводить рядки, в яких значення amount відсутня
new_orders = orders.copy() # створюємо копію таблиці
median_amnt = new_orders['amount'].median() # знаходимо медіану
new_orders["amount"] = new_orders["amount"].fillna(median_amnt) # заміщує пусті значення медіаною
print('Медіана = ', median_amnt) # виводить значення медіани
print('Кількість пустих значень:', new_orders['amount'].isna().sum()) # виводить кількість пустих значень по стовпцю amount
no_ampty_ord = orders.copy()
no_ampty_ord = no_ampty_ord.dropna(subset= 'amount')
print('Розмір таблиці без пустих значень amount (рядків, стовпців):',no_ampty_ord.shape)
```

```python
# Завдання 9. Зміна таблиці
work = orders.copy() # створює копію таблиці
work['amount'] = work['amount'] + 50 # збільшує всі значення в стовпці amount на 50
print("work['amount']=",work.loc['ord_1', 'amount']) #  виводить значення amount для ord_1 з обох таблиць
print("orders['amount']=",orders.loc['ord_1', 'amount']) 
work['avg_item'] = round(work['amount'] / work['items'], 2) # створює нову колонку в таблиці work
work = work.rename(columns={'source': 'channel'}) #  перейменовує колонку source на колонку channel в таблиці work
work.drop(columns={'express'}, inplace=True)# видаляє колонку express з таблиці work
print(work.columns) # виводить назви всіх колонок таблиці work
```

```python
# Завдання 10. Сортування та підсумки
print(orders.sort_values(by='amount', ascending = False)) # виводить таблицю відсортовану по стовпцю amount за спаданням
print(orders.sort_values(by=['source', 'amount'])) # сортує таблицю спочатку по стовпцю source, а потім, в розрізі кожного каналу, по amount та виводить результат
print('сума всіх значень колонки amount =',orders['amount'].sum()) # виводить суму всіх значень колонки amount
print('середнє значення колонки amount =',round(orders['amount'].mean(), 2)) # середнє значення стовпця amount за допомогою методу mean
#print('середнє арифметичне по колонці amount =',round(orders['amount'].sum()/len(orders), 2)) # середнє арифметичне по колонці amount
print('максимальне значення по колонці amount =',orders['amount'].max()) # вивоить рядки з максимальним значенням по колонці amount
print('мітка рядка з максимальним amount:',orders['amount'].idxmax()) #виводить мітку рядка з максимальним amount
```
