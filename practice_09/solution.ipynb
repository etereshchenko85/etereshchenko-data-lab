```python
# Завдання 1. Завантаження і перший огляд
import pandas as pd
import numpy as np

FILE = "Sample - Superstore Sales.xls"
orders = pd.read_excel(FILE, sheet_name="Orders")
returns = pd.read_excel(FILE, sheet_name="Returns")
users = pd.read_excel(FILE, sheet_name="Users")
print('Версія пандас',pd.__version__) # виводить версію Pandas
print('Розмір таблиці Orders:', orders.shape) # виводить розмір таблиці
print('Розмір таблиці Returns:', returns.shape) 
print('Розмір таблиці Users:', users.shape) 
#1.2
pd.set_option('display.max_columns', None) # налаштовує Pandas на відображення всіх стовпців в таблиці
#1.3
print(orders.head()) # виводить перші 5 (за замовчуванням) рядків таблиці
print(orders.tail()) # виводить останні 5 (за замовчуванням) рядків таблиці
#1.4
print('Розмір таблиці (рядків, ствопців):',orders.shape) # виводить розмір таблиці
print(orders.dtypes) # виводить типи даних кожного стовпця
#1.5
mb = orders.memory_usage(deep=True) # рахує, скільки пам'яті займає таблиця по стовпцях
print('Розмір таблиці в мегабайтах:',round(mb.sum()/1024**2, 2)) # Виводить загальний розмір таблиці
max_clmn = mb.idxmax() # знаходить стовпець, який займає набільше пам'яті
print("Найбільшу кіл-ть пам'яті займає стовпець:", max_clmn) 
```

```python
# Завдання 6. Робота з текстом
#6.1
orders_edit = orders.copy()
# додаємо колонки first_name та last_name та заповнюємо їх значенням з колонки Customer Name, розділяючи її по першому пробілу. аргумент expand визначає, чи сзберегти в нові колонки, чи повернути списком
orders_edit[['first_name', 'last_name']] = orders_edit['Customer Name'].str.split(n=1, expand=True) 
print('Кількість пустих значеь в стовпці з приізвищами:',orders_edit['last_name'].isna().sum()) #перевіряємо, чи є пусті значення в стовпці з прізвищами
#6.2
orders_edit['Name Length'] = orders_edit['Product Name'].str.len() # рахує кількість символів в кожній комірці стовпця та зберігає у новому стовпці
# виводить кількість знаків у найдовшій та найкоротшій назві товару
print('Найдовша назва товару:',orders_edit['Name Length'].max(),'\n',
      'Найкоротша назва товару:', orders_edit['Name Length'].min())
#6.3
#рахує та виводить кількість рядків, що мають в назві товару слово chair незалежно від регістру
print('Кількість рядків, що містять в назві товару Chair:', orders_edit['Product Name'].str.contains('Chair', case = False).sum())
#6.4
# Замінюємо значення в певних колонках 
orders_edit['Province'] = orders_edit['Province'].str.replace('Saskachewan', 'Saskatchewan', regex=False)
orders_edit['Product Sub-Category'] = orders_edit['Product Sub-Category'].str.replace('&', 'and', regex=False)
# виводимо унікальні значення з цих колонок
print(orders_edit['Province'].unique())
print(orders_edit['Product Sub-Category'].unique())
```

```python
# Завдання 7. Дати й час
#7.1
print(orders[['Order Date', 'Ship Date']].dtypes) # виводить тип даних по вказаних ствопцях
#7.2
orders2 = orders.copy()
orders2['Order Year'] = orders2['Order Date'].dt.year # створює стовпець з роком замовлення 
orders2['Order Month'] = orders2['Order Date'].dt.month  # створює стовпець з місяцем замовлення 
orders2['Order Weekday'] = orders2['Order Date'].dt.day_name() # створює стовпець з днем тижня замовлення 
orders2['Order Quarter'] = orders2['Order Date'].dt.quarter # створює стовпець з кварталом замовлення 
print(orders2.groupby('Order Year').size()) # групує по роках та рахує кількість рядків по кожному року
#7.3
orders2['Ship Delay'] = (orders2['Ship Date'] - orders2['Order Date']).dt.days # рахує різницю в днях між двома датами
# виводить мінімальне, максимальне та середнє значення по стовпцю Ship Delay
print(' Мінімальний термін оставки: ',orders2['Ship Delay'].min(), '\n', 
'Максимальний термін доставки : ',orders2['Ship Delay'].max(), '\n',
'Середній термін доставки :',round(orders2['Ship Delay'].mean()))
#7.4
# спочатку встановлює колонку Order Date як індекс, потім будує помісячний ряд та рахує суму продажів по кожному місяцю
month_sales = orders2.set_index('Order Date').resample('ME')['Sales'].sum()
print(f"Помісячний ряд складається з {round(len(month_sales), 2)} точок") # виводить кількість точок у ряді
#7.5
avg_delay = orders2.groupby('Order Weekday')['Ship Delay'].mean() # рахує середню затримку доставки по кожному дню тижня
print(round(avg_delay, 2))# виводить середню затримку доставки по кожному дню тижня
```

```python
# Завдання 8. Похідні колонки
#8.1
# створює одразу 3 нові стовпці методом assign, де аргументи - це нові стовпці
orders2 = orders2.assign(
    Profit_Margin = orders2['Profit'] / orders2['Sales'],
    Revenue_Per_Unit = orders2['Sales'] / orders2['Order Quantity'],
    Is_Profitable = orders2['Profit'] > 0
)
print(orders2.loc[:, ['Profit_Margin', 'Revenue_Per_Unit', 'Is_Profitable']]) # виводить створені стовпці
#8.2
#створює новий стовпець через лямбда функцію,  в яку передає значення Discount
orders2['Discount Level 1'] = orders2['Discount'].apply(
    lambda x: 'No' if x == 0 else('Low' if x < 0.05 else 'High')
)
# створює новий стовпець через розбиття на інтервали та призначення для них назв
orders2['Discount Level 2'] = pd.cut(
    orders2['Discount'],
    bins = [-float('inf'), 0, 0.04, float('inf')],
    labels = ['No', 'Low', 'High'],
    )
# порівнює два стовпці. через те, що різні методи повертають різні типи даних доводиться перетворювати Discount Level 2 на str
match = orders2['Discount Level 1'].equals(orders2['Discount Level 2'].astype(str))
print(match) # виводить результат порівняння
#8.3
# створюємо словник, де присвоюємо відповідний код кожному регіону
regions = {
    'West': 'W',
    'Prarie': 'PRA',
    'Atlantic': 'ATL',
    'Nunavut': 'NU',
    'Ontario': 'ON',
    'Northwest Territories': 'NWT',
    'Quebec' : 'QC',
    'Yukon': 'YT'
}
orders2['Region Code'] = orders2['Region'].map(regions) # робимо стовпець з відповідними кодами
print('Кількість пустих значень в колонці Region Code: ', orders2['Region Code'].isna().sum()) #перевіряємо на наявність пустих значень
#8.4
# створює новий стовпець через розбиття на інтервали та призначення для них назв
orders2['Delivery Speed'] = pd.cut(
    orders2['Ship Delay'],
    bins = [0, 3, 6, float('inf')],
    labels = ['швидке', 'звичайне', 'повільне'],
    right = False
)
print(orders2['Delivery Speed'].value_counts()) # виводить кількість значень в кожній категорії
#orders2
```

```python
# Завдання 9. Сортування та ранжування
#9.1
ord2sorted = orders2.sort_values(by='Sales', ascending=False) # сортує таблицю за спаданням по стовпцю Sales
print(ord2sorted['Sales'].iloc[0]) # виводить перше (найбільше) значення в стовпці Sales
# сортує таблицю спочатку по регіону за зростанням, а врозрізі кожного регіона по заробітку за спаданням
ord2sorted = orders2.sort_values(by=['Region', 'Profit'], ascending=[True, False])
#9.2
print(orders2['Profit'].nlargest(10)) # виводить 10 найбільших значень по стовпцю Profit
print(orders2['Profit'].nsmallest(10)) # виводить 10 найменших значень по стовпцю Profit
#9.3
# робить групування по регіонам та присвоює ранг продажам від найбільшого. равним значеннм присвоюється однаковий ранг
orders2['Sales Rank'] = orders2.groupby('Region')['Sales'].rank(ascending= False, method = 'dense')
# виводить строку з рангом 1 по регіону West
print(orders2.loc[
        (orders2['Region'] == 'West') &
        (orders2['Sales Rank'] == 1)]
      )
```

```python
# Завдання 10. Групування та агрегація
#10.1
# виводить дані, згруповані по категоріям продукту - суми по продажам та суми по заробітку
print(orders2.groupby('Product Category').agg({
    'Sales': 'sum',   
    'Profit': 'sum'  
})
)
#10.2
# створює нові колонки та виводить агреговані результати по ругіонам
print(orders2.groupby('Region').agg(
    total_sales = ('Sales', 'sum'),
    avg_discount = ('Discount', 'mean'),
    orders_count = ('Order ID', 'nunique'),
    max_profit = ('Profit', 'max')
    ))
#10.3
# групує дані по категоріям та підгатегоріям, рахує прибуток по кожній групі та виводить результат
grouped_categ = orders2.groupby(['Product Category', 'Product Sub-Category'])['Profit'].sum()
print(grouped_categ)
#10.4
# створює новий стовпець і для кожного рядка повертає середнє значення по категорії
orders2['Category Avg Sales'] = (orders2.groupby('Product Category')['Sales'].transform('mean'))
# створює новий стовпець і рахує різницю між поточним продажем та середнім по категорії
orders2['Sales Deviation'] = orders2['Sales'] - orders2['Category Avg Sales'] 
#10.5
# групуємо таблицю за категоріями та підкатегоріями, та обчислюємо суму прибутку. потім призначаємо нові індекси
categories = orders2.groupby(['Product Category', 'Product Sub-Category'])['Profit'].sum().reset_index()
# робимо нове групування по категоріям та визначаємо індекс з найбильшим прибутком
indx = categories.groupby('Product Category')['Profit'].idxmax()
# за допоогою iloc відбираємо рядки з найбільшим прибутком та виводимо результат
print(categories.iloc[indx])
```

```python
# Завдання 11. Зведені таблиці та зміна форми
#11.1
# будує зведену таблицю з orders2 з ітогами по рядках та стовпцях
piv_ord2 = pd.pivot_table(orders2, values='Sales', index='Region', columns ='Product Category', aggfunc='sum', margins=True) 
print(piv_ord2)
#11.2
print(pd.crosstab(orders2['Customer Segment'], orders2['Ship Mode'])) # створює кросс-таблицю по вказаних стовпцях 
#11.3
# один з варіантів отримати таблицю з першого завдання без ітогів. можна було створити наново без аргументу margins
pt_non_total = piv_ord2.drop('All').drop(columns='All') 
# перетворюємо Регіон з індекса на звичайний стовпець. а потім перетворюємо таблицю з широкого формату в довгий. 
m_piv_ord2 = pd.melt(pt_non_total.reset_index(), id_vars='Region', var_name='Product Category', value_name='Sales' )
print(f"В довогому форматі таблиці {len(m_piv_ord2)} рядки") # виводить кількість рядків в перетвореній таблиці (кількість регіонів * на кількість категорій )
```

```python
# Завдання 13. Бінінг, категорії та накопичувальні обчислення
# 13.1
orders2['Range_sales'] = pd.cut(orders2['Sales'], [0, 500, 2000, 10000, float('inf')], labels=['Low', 'Medium', 'High', 'Very High'])
print(orders2['Range_sales'].value_counts())
orders2['Equal_distr_sales'] = pd.cut(orders2['Sales'], bins=4, labels=['Part1', 'Part2', 'Part3', 'Part4'])
print(orders2['Equal_distr_sales'].value_counts())
# 13.2
orders2['q_profit'] = pd.qcut(orders2['Profit'], 4, labels=['Q1', 'Q2', 'Q3', 'Q4']) # розбиває стовпець Prifit на 4 рівні частини, кожні з них примвоює відповідну назву
print(orders2.groupby('q_profit', observed=True)['Sales'].mean()) # виводить згруповані по квартилях дані розраховуючи по кожному середні продажі
# 13.3
memory_us_bef = orders2[['Product Category', 'Region']].memory_usage(deep=True).sum() # вимірює обсяг пам'яті до перетворення
orders2['Product Category_cat'] = orders2['Product Category'].astype('category') # перетворює дані в тип категорія
orders2['Region_cat'] = orders2['Region'].astype('category')# перетворює дані в тип категорія
memory_us_aft = orders2[['Product Category_cat', 'Region_cat']].memory_usage(deep=True).sum() # вимірює обсяг пам'яті псля перетворення
print(f"Обсяг пам'яті до перетворення {memory_us_bef} байт, та після перетворення {memory_us_aft} байт") # виводить обсяг пам'яті до та після перетворення
print(f"Розмір змінився в {round(memory_us_bef/memory_us_aft)} рази") # виводить в скільки разів змінився обсяг використаної пам'яті
# 13.4
orders_sorted = orders2.sort_values('Order Date') # сортує таблицю по даті замовлення 
orders_sorted['Cumulative Sales'] = orders_sorted['Sales'].cumsum() #створює новий стовпець та рахує в ньому кумулятивну суму
print('Останнє значення Cumulative Sales:', orders_sorted['Cumulative Sales'].iloc[-1]) # виводить останнє (найбільше) значення в стовпці
# виводить проівняння дох значень. дані заокруглені задля уникнення відмінностей в числі після коми
print(round(orders_sorted['Sales'].sum(), 2) == round(orders_sorted['Cumulative Sales'].iloc[-1], 2))
#13.5
rolling_mean = month_sales.rolling(window=3).mean()  # рахує ковзне середнє з вікном у три місяці
expanding_mean = month_sales.expanding().mean() # рахує наростаюче середнє від початку періоду
pct_change = month_sales.pct_change() # рахує відсоткову зміну продажів між сусідніми місяцями
abs_change = month_sales.diff() # рахує абсолютну зміну продажів між сусідніми місяцями
print(rolling_mean, expanding_mean, pct_change, abs_change)
# виводить кількість пустих значень 
print(' rolling_mean',rolling_mean.isna().sum(), '\n',
      'expanding_mean', expanding_mean.isna().sum(), '\n',
      'pct_change', pct_change.isna().sum(), '\n',
      'abs_change', abs_change.isna().sum(), '\n')
```
